```python
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)['frame']
iris['target'] = ['setosa' if s==0 else 'versicolor' if s==1 else 'virginica' \
for s in iris['target']]
iris_melt = iris.melt(id_vars=['target'], var_name='variable', value_name='value')
(ggplot(iris_melt, aes(x='value'))
+ geom_histogram(bins=100)
+ facet_wrap('~variable'))
```

```python
(ggplot(iris_melt, aes(x='variable', y='value'))
 + geom_boxplot()
 + geom_jitter(alpha=0.1))
```

```python
(ggplot(iris_melt, aes(x='variable', y='value'))
 +geom_violin())
```
