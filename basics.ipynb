The `isna()` function tells which entries are `nan`.
Assign this logical vector to an object called `ind` and determine **how many NaNs** the `age` column has.

4. Use search or documentation to determine **how to compute the mean** of only the present (non-NaN) values of the `age` vector.


````python
import seaborn as sns
import numpy as np

fahrenheit_temp = np.array([35, 88, 42, 84, 81, 30])
celsius_temp = 5 / 9 * (fahrenheit_temp - 32)
print(celsius_temp)

titanic = sns.load_dataset("titanic")
count = np.count_nonzero(titanic['age'].isna())
print(count)

mean = np.nanmean(titanic["age"])
print(mean)
````
