# Section 03 - Manipulating data tables
1. Add a new column called High_Rating to the DataFrame ratings_df. The column has an integer
1 for all observations with a rating value higher than 7. Hint: You can create a boolean variable and

```python
import os
import pandas as pd

ratings_file = os.path.join("./BX-Book-Ratings.csv")
users_file = os.path.join("./BX-Users.csv")

ratings_df = pd.read_csv(ratings_file)
users_df = pd.read_csv(users_file)

print(ratings_df)
print(users_df)

ratings_df["High_Rating"] = (ratings_df["Book-Rating"] > 7).astype(int) # 1.
print(ratings_df)

str((ratings_df["High_Rating"].sum() / ratings_df["High_Rating"].count() * 100).round(4)) + "% " # 2.

subset_users_df = users_df[users_df["User-ID"].isin(ratings_df["User-ID"])] # 3.
print(subset_users_df)

print(round(subset_users_df["Age"].mean(), 3), "is the common age of users who rated a book") # 4.

print(round(ratings_df["ISBN"].count() / subset_users_df["User-ID"].count(), 3), "Books per user") # 5.

average_rating_df = ratings_df.groupby('ISBN', as_index=False).agg(average_rating=('Book-Rating', 'mean'), publishing_year=('Year-Of-Publication', 'first'), author=('Book-Author', 'first'),review_count=('ISBN', 'count'), max_rating = ('Book-Rating', 'max')) # 8.
average_rating_df = average_rating_df[average_rating_df['publishing_year'] != 0]
print(average_rating_df)
maximum_rating_df = average_rating_df[
          (average_rating_df['average_rating'] == average_rating_df['average_rating'].max())
      ]
print(maximum_rating_df[maximum_rating_df['publishing_year'] == maximum_rating_df['publishing_year'].min()], "is the earliest max rated book") # 6.

print(average_rating_df[average_rating_df['review_count'] == average_rating_df['review_count'].max()], "is the book with the most reviews") # 7.

authors = ["Agatha Christie", "William Shakespeare", "Stephen King",
"Ann M. Martin", "Carolyn Keene", "Francine Pascal",
"Isaac Asimov", "Nora Roberts", "Barbara Cartland", "Charles Dickens"]
chosen_authors_df = average_rating_df[average_rating_df['author'].isin(authors)] # 9.
print(chosen_authors_df)

authors_by_ratings = chosen_authors_df.groupby('author', as_index=False).agg(total_ratings=('review_count', 'sum')) # 10.
print(authors_by_ratings)
```
